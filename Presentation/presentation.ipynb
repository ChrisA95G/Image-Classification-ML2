# **Unveiling Cellular Secrets: The Human Protein Atlas Image Classification Challenge**

Dataset: Human Protein Atlas Image Classification.

*Niklas Hohn, Christopher Arleth*

## **The Challenge:**
* To classify the location(s) of proteins within a cell from microscope images.
* This is a multi-label image classification task, as a single protein can be found in multiple locations.

## **Images:** What are we looking at?
* Cell images from (confocal microscopy)
    * Protein of interest (the target for classification)
    * Nucleus
    * Microtubules
    * Endoplasmic Reticulum

## **What we did:**
* Our approach:
    - We used a modified Visual image Transformer (ViT) with a custom architecture.
    - Data augmentation techniques were applied to improve model generalization.
    - We utilized transfer learning from pre-trained models to enhance performance.
* Results:
    - Our model achieved an F1: 0.744 on the validation set.
    - The model was able to classify multiple labels per image effectively (Exact matches 59.5 %).
